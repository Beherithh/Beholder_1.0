```python
import ccxt

exchange = ccxt.mexc()
markets = exchange.load_markets()
ticket = "AKE"

symbol = f'{ticket}/USDT:USDT'
market = exchange.market(symbol)

print(f"Тип рынка: {market['type']}")               # 'swap'
print(f"Размер 1 контракта: {market['contractSize']}") # например, 0.0001
```

```python
# Запрос последних 3 часовых свечей
ohlcv = exchange.fetch_ohlcv(symbol, timeframe='1d', limit=3)

for candle in ohlcv:
    timestamp, open_p, high_p, low_p, close_p, volume_contracts = candle

    # Перевод объёма из контрактов в базовую валюту
    volume_in_coins = volume_contracts * market['contractSize']

    print(f"Время: {exchange.iso8601(timestamp)}")
    print(f"Объём в контрактах: {volume_contracts}")
    print(f"Объём в {ticket}: {volume_in_coins}")
    print(f"объём в USDT {volume_in_coins * close_p}\n")
```
